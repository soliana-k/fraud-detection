## Class imbalance handling - SMOTE

The creditcard dataset is severely imbalanced fraudulent transactions account for
roughly 0.17% of all records. A naive classifier that predicts non-fraud for every
transaction would achieve over 99% accuracy while completely failing its purpose.

SMOTE (Synthetic Minority Oversampling Technique) is used to oversample the minority
fraud class in the training set. It was preferred over undersampling because the dataset,
while large, would lose substantial majority-class information under aggressive
undersampling. SMOTE generates synthetic fraud samples along the feature-space
boundaries of existing fraud cases, giving the model more signal to learn the fraud
pattern without simply duplicating records.

SMOTE is applied **after the train/test split and only to training data** to prevent
leakage of any synthetic signal into evaluation.

## Class imbalance handling - SMOTE

The dataset is highly imbalanced, with fraudulent transactions representing a small
minority of the data. Training a model on imbalanced data causes it to be biased
toward the majority class (non-fraud), resulting in high accuracy but poor recall
on the fraud class which is the class that matters most.

SMOTE (Synthetic Minority Oversampling Technique) was chosen over random undersampling
because undersampling discards legitimate majority-class examples, which wastes
information. SMOTE instead generates synthetic minority-class samples by interpolating
between existing fraud examples, preserving all majority-class data while balancing
the distribution.

Critically, SMOTE is applied **to the training set only** after the train/test split.
Applying it before splitting would cause data leakage synthetic samples derived from
the test set would bleed into training, artificially inflating model performance metrics.
